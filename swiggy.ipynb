```python
categorical_nom = ['weather',
 'type_of_order',
 'type_of_vehicle',
 'festival', 
 'city_name',
 'order_day_of_week']

categorical_ord = [
'traffic','city_type','order_time_of_day']
```

# Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
numerical_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
categorical_nom_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('encoder', OneHotEncoder())
])
```

```python
categorical_ord_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('encoder', OrdinalEncoder())
])
```

```python
Preprocessor = ColumnTransformer(transformers =[
    ('num_pipe',numerical_pipeline,numerical),
    ('cat_nom_pipe',categorical_nom_pipeline,categorical_nom),
    ('cat_ord_pipe',categorical_ord_pipeline,categorical_ord)
])
```

```python
Preprocessor
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
```

```python
model_lin_reg = Pipeline([
    ('preprocess',Preprocessor),
    ('model_lr',LinearRegression())
])
model_knn_reg = Pipeline([
    ('preprocess',Preprocessor),
    ('model_knn',KNeighborsRegressor())
])
model_dt_reg = Pipeline([
    ('preprocess',Preprocessor),
    ('model_dt',DecisionTreeRegressor())
])
model_rf = Pipeline([
    ('preprocess',Preprocessor),
    ('model_rf',RandomForestRegressor(random_state=42))
])
model_gb_reg = Pipeline([
    ('preprocess',Preprocessor),
    ('model_gb',GradientBoostingRegressor(random_state=42))
])
model_xg = Pipeline([
    ('preprocess',Preprocessor),
    ('model_xg',XGBRegressor(random_state=42))
])
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state=42)
```

# Linear Regression

```python
model_lin_reg
```

```python
model_lin_reg.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
```

```python
y_pred = model_lin_reg.predict(X_test)

r2 = r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mean_squared_error(y_test,y_pred))

print("R2 score: ",r2)
print("MAE: ",mae)
print("RMSE: ",rmse)
```

# KNN Regressor

```python
model_knn_reg.fit(X_train,y_train)
```

```python
y_pred = model_knn_reg.predict(X_test)

r2 = r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mean_squared_error(y_test,y_pred))

print("R2 score: ",r2)
print("MAE: ",mae)
print("RMSE: ",rmse)
```

```python
models = {
    "Decision Tree": model_dt_reg,
    "Random Forest": model_rf,
    "Gradient Boosting": model_gb_reg,
    "XGBoost": model_xg
}
```

```python
for name, model in models.items():
    print(f"Training {name}...")

    model.fit(X_train, y_train)

    print(f"{name} completed!")

    y_pred = model.predict(X_test)

    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))

    print(f"R2 Score: {r2:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"RMSE: {rmse:.4f}\n")
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
pipe = Pipeline([
    ('preprocess',Preprocessor),
    ('model',XGBRegressor(random_state=42))
])

param_dist = {
    'model__n_estimators': [100, 200, 300, 500],
    'model__learning_rate': [0.01, 0.05, 0.1, 0.2],
    'model__max_depth': [3, 5, 7, 9],
    'model__subsample': [0.7, 0.8, 0.9, 1.0],
    'model__colsample_bytree': [0.7, 0.8, 0.9, 1.0]
}

search = RandomizedSearchCV(
    estimator = pipe,
    param_distributions=param_dist,
    n_iter = 20,
    cv = 5,
    scoring = 'r2',
    random_state=42,
    n_jobs = -1
)

search.fit(X_train,y_train)

print("Best Parameters:", search.best_params_)
print("Best CV Score:", search.best_score_)
```

```python
# Training score
train_pred = search.predict(X_train)
train_r2 = r2_score(y_train, train_pred)

# Test score
test_pred = search.predict(X_test)
test_r2 = r2_score(y_test, test_pred)

print("Train R²:", train_r2)
print("Test R²:", test_r2)
```
